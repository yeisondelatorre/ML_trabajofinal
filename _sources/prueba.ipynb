```python
import tensorflow as tf

# Verificar cuántas GPUs están disponibles
gpus = tf.config.experimental.list_physical_devices('GPU')
if gpus:
    # Habilitar el crecimiento de memoria en la GPU
    for gpu in gpus:
        tf.config.experimental.set_memory_growth(gpu, True)
    print(f"Se ha detectado {len(gpus)} GPU(s):")
    for gpu in gpus:
        print(f"  {gpu.name}")
else:
    print("No se ha detectado ninguna GPU.")


```

```python
import tensorflow as tf

# Verificar si TensorFlow detecta la GPU
gpus = tf.config.experimental.list_physical_devices('GPU')
if gpus:
    print(f"Se ha detectado {len(gpus)} GPU(s):")
    for gpu in gpus:
        print(f"  {gpu.name}")
    # Configurar para que TensorFlow use la GPU
    for gpu in gpus:
        tf.config.experimental.set_memory_growth(gpu, True)
else:
    print("No se ha detectado ninguna GPU.")
```
